### Numpy Array vs Python Lists
Numpy Array are:
1. Faster
2. Convienent
3. Less Memory

```python
py_list = range(100)
np_arr = np.arange(100)
```

#### Memory Comparison

```python
# Memory comparison between python list and numpy array
import sys
ONE_MB = 1024 * 1024

mem_python_list = sys.getsizeof(85) * len(py_list)
mem_numpy_arr = np_arr.size * np_arr.itemsize
print(f"memory for 100 item python list {mem_python_list} bytes")
print(f"memory for 100 item numpy array {mem_numpy_arr} bytes")
print(f"100K items: memory -> python list: {(100000 * mem_python_list)/ONE_MB} MB")
print(f"100K items: memory -> numpy arr: {(100000 * mem_numpy_arr)/ONE_MB} MB")
```

#### Efficiency Comparison

```python
import time
x = range(100000)
y = range(100000, 200000)

start_time = time.time() # start time
result = [x + y for x, y in zip(x, y)]
end_time = time.time() # end time
print(f"Time Spent: {end_time-start_time} seconds")
```

```python
x1 = np.arange(100000)
y1 = np.arange(100000, 200000)
start_time = time.time() # start time
result = x1 + y1
end_time = time.time() # end time
print(f"Time Spent: {end_time-start_time} seconds")
```

### Reshaping Numpy Arrays
We can use following method to change the shape of the ndarrays. Here are 5 methods:
1. Ravel -> convert higher dimensional arrays into 1D arrays.
2. Reshape -> changes the shape of the array to other dimensions.
3. Transpose -> Column to Row and vice versa
4. Stacking -> Combine two same shape arrays using following functions:
- hstack
- vstack
5. Splitting -> Split array into multiple arrays
- hsplit
- vsplit

```python
# Ravel is similar to nditer.
arr = np.arange(24).reshape(4, 6)
print("Before Ravel operations")
print(f"arr: {arr} has shape {arr.shape} and is {arr.ndim} dimension")
arr_ravel = arr.ravel()
print("After Ravel operations")
print(f"Ravel arr: {arr_ravel} has shape {arr_ravel.shape} and is {arr_ravel.ndim} dimension")
```

```python
# Transpose
print(f"Original array:\n {arr}")
print(f"Transpose array:\n {arr.transpose()}")
```

```python
# Stacking --> hstack and vstack
arr5 = np.arange(0, 6).reshape(2, 3)
arr6 = np.arange(6, 12).reshape(2,3)
print("Before stacking")
print(f"arr5: {arr5}")
print()
print(f"arr6: {arr6}")
print("\n Horizontal Stacking")
print(np.hstack((arr5, arr6)))
print("\n Vertical Stacking")
print(np.vstack((arr5, arr6)))
```

### Indexing with boolean arrays

```python
arr = np.random.randint(low=1, high=100, size=20).reshape(4, 5)
arr
```

```python
# Generate boolean array from array.
# In this case we will use comparison operators to do element wise comparison.
arr > 50
```

```python
# indexing using Boolean arrays - aka Filtering
print(f"Output element > 50: {arr[arr>50]}")
print(f"Output element > 50 and Odd nubers: {arr[(arr>50) & (arr%2 != 0)]}")
print(f"Set element > 50 and Odd nubers to 0")
arr[(arr>50) & (arr%2 != 0)] = 0
```
