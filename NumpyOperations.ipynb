### **Creating Arrays**
There are 6 different ways to create NumPy arrays:
1. Using np.array()
2. Using np.zero(), np.ones(), np.empty(), np.random()
3. Using np.arange()
4. Uing np.linspace()
5. Using Copy()
6. Using Identity()

```python
import numpy as np
```

```python
# Generates a vector
one_dim = [1,2,3,4,5]
arr = np.array(one_dim)
arr
```

```python
print(type(arr))
```

```python
# Generates a 2D matrix
two_dim = [[1,2,3,4,5], [6,7,8,9, 0]]
two_dim_ndarray = np.array(two_dim)
two_dim_ndarray
```

```python
# Generates matrix of 0's
zeros = np.zeros((2,3))
zeros
```

```python
# Generates matrix of 1's
ones = np.ones((2,3))
ones
```

```python
# Generates an identity matrix
identity = np.identity(5)
identity
```

```python
# Generates an array in the range of [start, stop)
arr_range = np.arange(1, 10)
arr_range
```

```python
# Generates an array/matrix of equidistant points or Returns evenly spaced numbers
equi_array = np.linspace(10, 20, 15)
equi_array
```

```python
# Generates a copy of the numpy array
identity_copy = identity.copy()
identity_copy
```

###**Properties & Attributes**
Numpy array is a data type and will have properties and attributes. Some of these are listed here:
1. shape - reutrns a tuple of array dimension.
2. ndim
3. size
4. itemsize
5. dtype
6. astype()

```python
print(f"Array {arr} has shape {arr.shape}")
print(f"Dimension: {arr.ndim} dimension matrix")
print(f"Size: {arr.size}")
print(f"ItemSize: {arr.itemsize}")
print(f"Dtype: {arr.dtype}")
```

```python
print(f"Array {two_dim_ndarray} has shape {two_dim_ndarray.shape}")
print(f"Dimension: {two_dim_ndarray.ndim} dimension matrix")
print(f"Size: {two_dim_ndarray.size}")
print(f"ItemSize: {two_dim_ndarray.itemsize}")
print(f"Dtype: {two_dim_ndarray.dtype}")
```

```python
arr_3d = np.array([[[1,3], [5, 7]], [[2, 4], [4, 6]]])
print(f"Array {arr_3d} has shape {arr_3d.shape}")
print(f"Dimension: {arr_3d.ndim} dimension matrix")
print(f"Size: {arr_3d.size}")
print(f"ItemSize: {arr_3d.itemsize}")
print(f"Dtype: {arr_3d.dtype}")
```

```python
# Convert data type of numpy array
print(f"Before converting data type: {arr_3d}")
arr_3d = arr_3d.astype('float')
print(f"After converting data type: {arr_3d}")
```

### Numpy Array vs Python Lists
Numpy Array are:
1. Faster
2. Convienent
3. Less Memory

```python
py_list = range(100)
np_arr = np.arange(100)
```

#### Memory Comparison

```python
# Memory comparison between python list and numpy array
import sys
ONE_MB = 1024 * 1024

mem_python_list = sys.getsizeof(85) * len(py_list)
mem_numpy_arr = np_arr.size * np_arr.itemsize
print(f"memory for 100 item python list {mem_python_list} bytes")
print(f"memory for 100 item numpy array {mem_numpy_arr} bytes")
print(f"100K items: memory -> python list: {(100000 * mem_python_list)/ONE_MB} MB")
print(f"100K items: memory -> numpy arr: {(100000 * mem_numpy_arr)/ONE_MB} MB")
```

#### Efficiency Comparison

```python
import time
x = range(100000)
y = range(100000, 200000)

start_time = time.time() # start time
result = [x + y for x, y in zip(x, y)]
end_time = time.time() # end time
print(f"Time Spent: {end_time-start_time} seconds")
```

```python
x1 = np.arange(100000)
y1 = np.arange(100000, 200000)
start_time = time.time() # start time
result = x1 + y1
end_time = time.time() # end time
print(f"Time Spent: {end_time-start_time} seconds")
```
